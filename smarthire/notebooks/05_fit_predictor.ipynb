```python
import pandas as pd
import sys
sys.path.append('../')
from src.models.fit_predictor import calculate_fit_score
```

```python
candidates_data = [
    {"Applicant": "Candidate A", "Cosine_Sim": 0.38, "Missing_Skills": 0},
    {"Applicant": "Candidate B", "Cosine_Sim": 0.25, "Missing_Skills": 4},
    {"Applicant": "Candidate C", "Cosine_Sim": 0.42, "Missing_Skills": 1},
    {"Applicant": "Candidate D", "Cosine_Sim": 0.15, "Missing_Skills": 7},
    {"Applicant": "Candidate E", "Cosine_Sim": 0.31, "Missing_Skills": 2}
]
df_candidates = pd.DataFrame(candidates_data)
```

```python
df_candidates['Final_Fit_Score_%'] = df_candidates.apply(
    lambda row: calculate_fit_score(row['Cosine_Sim'], row['Missing_Skills']), axis=1
)
shortlist = df_candidates.sort_values(by='Final_Fit_Score_%', ascending=False).reset_index(drop=True)
print("Candidate Shortlisting Comparison Table:")
display(shortlist)
```
