```python
#Importing necessary libraries and loading the dataset
import pandas as pd
import numpy as np

# 1.Load the dataset using Pandas.
df = pd.read_csv('student_scores.csv')
```

```python
# Part 1 — NumPy Operations
# 1.Create a NumPy array from the math_score column.
math_scores_raw = df['math_score'].values

# 2.Find the mean, median, maximum, and minimum values.
mean_val = np.nanmean(math_scores_raw)
median_val = np.nanmedian(math_scores_raw)
max_val = np.nanmax(math_scores_raw)
min_val = np.nanmin(math_scores_raw)

print(f"Mean Math Score: {mean_val:.2f}")
print(f"Median Math Score: {median_val:.2f}")
print(f"Maximum Math Score: {max_val}")
print(f"Minimum Math Score: {min_val}")

# 3.Normalize the scores using NumPy.
normalized_scores = (math_scores_raw - min_val) / (max_val - min_val)
print("\nNormalized Math Scores Array (First 5 records):")
print(normalized_scores[:5])
```

```python
# Part 2 — Pandas Exploration

# 2.Display the first 5 rows of the dataset.
print("1. First 5 Rows of the Dataset:")
print(df.head())

# 3.Check the data types of all columns.
print("\n2. Column Data Types:")
print(df.dtypes)

# 4.Find the total number of missing values in each column
print("\n3. Missing Values count per Column:")
print(df.isnull().sum())

# 5.Find students with attendance below 70%.
print("\n4. Students with Attendance Below 70%:")
print(df[df['attendance'] < 70][['student_id', 'name', 'attendance']])
```

```python
# Part 3 — Data Preprocessing
# 1.Identify and handle missing values.
df_cleaned = df.drop_duplicates().copy()

# 2.Convert incorrect data formats into proper formats.
df_cleaned['age'] = pd.to_numeric(df_cleaned['age'].replace('twenty', 20))
df_cleaned['science_score'] = df_cleaned['science_score'].replace(999.0, np.nan)
df_cleaned['math_score'] = df_cleaned['math_score'].replace(150.0, np.nan)

# 3.Convert the exam_date column into datetime format.
df_cleaned['exam_date'] = pd.to_datetime(df_cleaned['exam_date'], errors='coerce')

# 4.Detect and handle outliers in math_score and science_score columns.
num_cols = ['age', 'math_score', 'science_score', 'attendance']
df_cleaned[num_cols] = df_cleaned[num_cols].fillna(df_cleaned[num_cols].median())
df_cleaned['gender'] = df_cleaned['gender'].fillna(df_cleaned['gender'].mode()[0])

# 5.Find and remove duplicate rows.
print("\nMissing Values After Preprocessing Imputation:")
print(df_cleaned.isnull().sum().drop('exam_date'))
```

```python
# Part 4 — Data Analysis
# 1.Create a new column named average_score.
df_cleaned['average_score'] = (df_cleaned['math_score'] + df_cleaned['science_score']) / 2

# 2.Find the top 5 students based on average_score.
print("\nTop 5 Students Based on Average Score:")
print(df_cleaned.sort_values(by='average_score', ascending=False).head(5)[['name', 'math_score', 'science_score', 'average_score']])

# 3.Calculate the correlation between attendance and marks.
print("\nCorrelation Matrix (Attendance vs. Marks):")
print(df_cleaned[['attendance', 'math_score', 'science_score', 'average_score']].corr())

# 4.Group students by gender and calculate average marks.
print("\nAverage Marks Grouped by Gender:")
print(df_cleaned.groupby('gender')[['math_score', 'science_score', 'average_score']].mean())

df_cleaned.to_csv('cleaned_student_scores.csv', index=False)
```
